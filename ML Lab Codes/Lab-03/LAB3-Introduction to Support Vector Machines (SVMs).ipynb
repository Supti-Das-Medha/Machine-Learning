```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.tree import DecisionTreeClassifier, plot_tree
random_state = 272
np.random.seed(random_state )
print("SUCESSFULLY IMPORTED")
```

```python
df = pd.read_csv('heart_disease.csv')
df.head()
```

```python
print(df.sample())
```

```python
print(df.info())
```

```python

print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df["RestingBP"] = df["RestingBP"].fillna(df["RestingBP"].median())
df["Cholesterol"] = df["Cholesterol"].fillna(df["Cholesterol"].median())
df["MaxHR"] = df["MaxHR"].fillna(df["MaxHR"].median())
print(df.isnull().sum())
```

```python
X = df[['Age','RestingBP','FastingBS','MaxHR','Oldpeak']]

y = df['HeartDisease']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=random_state)
```

```python
print(f"X-train: {X_train.shape}")
print(f" X_test: { X_test.shape}")
print(f"y_train: {y_train.shape}")
print(f" y_test: { y_test.shape}")
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test =  scaler.transform(X_test)

print("\n Features scaled sicessfully")
```

```python
SVM_2d_linear_new = SVC(kernel = 'linear', random_state = 42)
SVM_2d_linear_new.fit(X_train, y_train)
model = SVC(kernel='linear')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred,y_test)
print("Evaluate the performance of the Linear SVC model")
print(classification_report(y_test,y_pred))
print(accuracy)
print(confusion_matrix(y_test, y_pred))
```

```python
SVM_2d_linear_new = SVC(kernel = 'rbf', random_state = 42)
SVM_2d_linear_new.fit(X_train, y_train)
model = SVC(kernel='rbf')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred,y_test)
print("Evaluate the performance of the RBF SVC model")
print(classification_report(y_test,y_pred))
print(accuracy)
print(confusion_matrix(y_test, y_pred))
```

```python
SVM_2d_linear_new = SVC(kernel = 'poly', random_state = 42)
SVM_2d_linear_new.fit(X_train, y_train)
model = SVC(kernel='poly')
model.fit(X_train,y_train)
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_pred,y_test)

print("Evaluate the performance of the Polynomial SVC model")
print(classification_report(y_test,y_pred))
print(accuracy)
print(confusion_matrix(y_test, y_pred))
```

```python
clf = DecisionTreeClassifier(criterion='entropy', max_depth=3, random_state=42)
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
accuracy = accuracy_score(y_pred,y_test)
print("Accuracy using Decision Tree:", accuracy)
```
